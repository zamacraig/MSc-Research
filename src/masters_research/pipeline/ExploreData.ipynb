```python
attribute_df.head(20)
```

```python
# 1. Prepare the Date column
attribute_df['Date'] = pd.to_datetime(
    attribute_df['Year'].astype(str) + '-' + attribute_df['Month'].astype(str) + '-01'
)

# 2. Add 'Presence' flag and Pivot
attribute_df['Presence'] = 1
presence_pivot = attribute_df.pivot_table(
    index='Compno', 
    columns='Date', 
    values='Presence', 
    aggfunc='max'
).fillna(0)

# 3. Set up the visualization
plt.figure(figsize=(16, 8)) # Increased height slightly for better label spacing
custom_cmap = ListedColormap(['#f4f4f4', 'forestgreen'])
```

```python
# Draw the heatmap
ax = sns.heatmap(
    presence_pivot, 
    cmap=custom_cmap, 
    cbar=False,            
    linewidths=0
)

# 4. Formatting the plot
plt.title('Data Presence per CompNo (2000-2017)', fontsize=16, pad=20)
plt.xlabel('Year', fontsize=12)
plt.ylabel('CompNo', fontsize=12)

# --- READABLE X-AXIS (Angled Labels) ---
dates = presence_pivot.columns
# Position a tick only for the first month of every year
tick_positions = [i + 0.5 for i, d in enumerate(dates) if d.month == 1]
tick_labels = [d.strftime('%Y') for d in dates if d.month == 1]

ax.set_xticks(tick_positions)
# 'rotation=45' tilts the labels, 'ha=right' aligns them correctly to the tick
ax.set_xticklabels(tick_labels, rotation=45, ha='right') 

# --- HORIZONTAL Y-AXIS ---
# 'rotation=0' ensures CompNo labels are horizontal
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

plt.tight_layout()
plt.show()
```
